# Classification dataset loader check

Smoke-test every loader listed in `configs/datasets/classification_all.json` against a fresh cache folder, so each dataset is downloaded from the remote source rather than served from an existing cache. Useful for spotting datasets that have become private / unreachable.

Two equivalent variants are shown:

1. **Direct** — call `raman_data(name, cache_dir=...)` for each key (one cell per dataset).
2. **Benchmark** — instantiate `RamanBenchmark` with `cache_dir=...` and let it manage the cache layout.

```python
import json
from pathlib import Path
from raman_data import raman_data

%load_ext autoreload
%autoreload 2

REPO_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATASETS_FILE = REPO_ROOT / "configs" / "datasets" / "classification_all.json"

CACHE_DIR = REPO_ROOT / ".cache_loader_check"
CACHE_DIR.mkdir(parents=True, exist_ok=True)

# This is only to align with our RamanBenchmark object (see last cell)
raw_cache = CACHE_DIR / "datasets_raw"
raw_cache.mkdir(exist_ok=True)

keys = json.loads(DATASETS_FILE.read_text())
print(f"Cache dir: {CACHE_DIR}")
print(f"{len(keys)} classification datasets:")
for key in keys:
    print(f"  {key}")
```

## Variant 2: via `RamanBenchmark`

Uses the benchmark wrapper, which manages the raw/processed cache layout under the same `cache_dir` and applies preprocessing + train/test splitting.

```python
from raman_bench import RamanBenchmark

bench = RamanBenchmark(
    dataset_names_classification=keys,
    dataset_names_regression=[],
    cache_dir=str(CACHE_DIR),
    min_samples_per_class=9,
    seed=0,
)
bench.init_datasets()
print(f"\nPrepared splits: {len(bench)} / {len(keys)}")
```
